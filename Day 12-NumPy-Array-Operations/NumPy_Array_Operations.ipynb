# NumPy Array Operations

## Day 12 - AI & ML Internship Task

This project demonstrates basic mathematical operations on NumPy arrays.

### Objective

- Create two NumPy arrays
- Perform addition, subtraction, multiplication, and division
- Check the shape of both arrays
- Understand vectorized operations using NumPy

```python
import numpy as np

# Create two NumPy arrays
array1 = np.array([10, 20, 30, 40, 50])
array2 = np.array([2, 4, 5, 8, 10])

print("Array 1:", array1)
print("Array 2:", array2)
```

```python
# NumPy array operations

addition = array1 + array2
subtraction = array1 - array2
multiplication = array1 * array2
division = array1 / array2

print("Addition:", addition)
print("Subtraction:", subtraction)
print("Multiplication:", multiplication)
print("Division:", division)
```

```python
# Check the shape of both arrays

print("Shape of Array 1:", array1.shape)
print("Shape of Array 2:", array2.shape)
```

## Explanation

NumPy supports vectorized operations, which means mathematical operations can be performed directly on entire arrays without writing separate loops for each element.

In this task, addition, subtraction, multiplication, and division were performed element-by-element between the two arrays.

The `shape` property shows the dimensions of an array. Since both arrays contain 5 elements, their shape is `(5,)`.

Vectorized operations make numerical calculations faster, simpler, and easier to work with, especially when handling large datasets in data analysis and machine learning.

## Interview Questions

### 1. What is vectorization in NumPy?

Vectorization is the process of performing operations on entire NumPy arrays without using explicit loops. It makes calculations simpler and usually faster.

### 2. What is the difference between a Python list and a NumPy array?

A Python list can store different types of data and is mainly used for general-purpose programming. A NumPy array is designed for numerical calculations and supports fast mathematical operations.

### 3. Why is NumPy widely used in Machine Learning?

NumPy provides efficient arrays and mathematical operations that are useful for handling numerical data. Many data science and machine learning libraries also use NumPy internally.
