```python
# Student Grade Calculator

students = {
    "Swetha": 85,
    "Rahul": 72,
    "Priya": 91,
    "Arjun": 64,
    "Anjali": 48
}

for name, marks in students.items():
    if marks >= 90:
        grade = "A+"
    elif marks >= 80:
        grade = "A"
    elif marks >= 70:
        grade = "B"
    elif marks >= 60:
        grade = "C"
    elif marks >= 50:
        grade = "D"
    else:
        grade = "F"

    print(name, ":", marks, "→ Grade", grade)
```
