```python
# Text Analysis

text = """
Python is a powerful programming language. Python is easy to learn and Python is widely used in AI and Machine Learning.
"""

print("Text:")
print(text)
```

```python
# Count words and characters

words = text.split()

word_count = len(words)
character_count = len(text)

print("Number of Words:", word_count)
print("Number of Characters:", character_count)
```

```python
from collections import Counter

# Count vowels
vowels = "aeiou"
vowel_count = 0

for character in text.lower():
    if character in vowels:
        vowel_count += 1

# Find repeated words
words_clean = [word.strip(".,").lower() for word in text.split()]
word_frequency = Counter(words_clean)

repeated_words = {}

for word, count in word_frequency.items():
    if count > 1:
        repeated_words[word] = count

print("Number of Vowels:", vowel_count)
print("Repeated Words:", repeated_words)
```
