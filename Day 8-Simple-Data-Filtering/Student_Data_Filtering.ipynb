```python
# Simple Data Filtering

students = [
    {"name": "Swetha", "marks": 85, "age": 20, "department": "CSE"},
    {"name": "Rahul", "marks": 72, "age": 21, "department": "ECE"},
    {"name": "Priya", "marks": 91, "age": 20, "department": "CSE"},
    {"name": "Arjun", "marks": 64, "age": 22, "department": "EEE"},
    {"name": "Anjali", "marks": 78, "age": 21, "department": "CSE"},
    {"name": "Kiran", "marks": 88, "age": 20, "department": "ECE"}
]

print("Student Records:")
for student in students:
    print(student)
```

```python
# Filter students with marks 80 or above

high_mark_students = []

for student in students:
    if student["marks"] >= 80:
        high_mark_students.append(student)

print("Students with marks 80 or above:")
for student in high_mark_students:
    print(student)
```

```python
# Filter students from CSE department

cse_students = []

for student in students:
    if student["department"] == "CSE":
        cse_students.append(student)

print("Students from CSE Department:")
for student in cse_students:
    print(student)
```

```python
# Filter CSE students with marks 80 or above

filtered_students = []

for student in students:
    if student["marks"] >= 80 and student["department"] == "CSE":
        filtered_students.append(student)

print("CSE Students with marks 80 or above:")
for student in filtered_students:
    print(student)
```
