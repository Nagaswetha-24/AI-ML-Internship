```python
# Python Data Processing

records = [
    {"name": "Swetha", "department": "CSE", "marks": 85},
    {"name": "Rahul", "department": "ECE", "marks": 72},
    {"name": "Priya", "department": "CSE", "marks": 91},
    {"name": "Arjun", "department": "EEE", "marks": 64},
    {"name": "Anjali", "department": "CSE", "marks": 78}
]

print("Student Records:")
for record in records:
    print(record)
```

```python
# Summary Statistics

total_marks = sum(record["marks"] for record in records)
average_marks = total_marks / len(records)
highest_marks = max(record["marks"] for record in records)
lowest_marks = min(record["marks"] for record in records)

print("----- Summary Statistics -----")
print("Number of Students:", len(records))
print("Total Marks:", total_marks)
print("Average Marks:", average_marks)
print("Highest Marks:", highest_marks)
print("Lowest Marks:", lowest_marks)
```

```python
# Function to process a student record

def process_record(record):
    result = {
        "name": record["name"],
        "marks": record["marks"],
        "status": "Pass" if record["marks"] >= 50 else "Fail"
    }
    return result

print("----- Processed Records -----")

for record in records:
    print(process_record(record))
```

```python
# Filter students with marks 80 or above

high_scorers = []

for record in records:
    if record["marks"] >= 80:
        high_scorers.append(record)

print("----- Students with Marks 80 or Above -----")

for record in high_scorers:
    print(record)
```
